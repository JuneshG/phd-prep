(c) $\sum_{j=1}^{3} a_{2j}x_{j}$ for ${A}$ = $\begin{bmatrix}1 & 2 & 0 \\3 & -1 & 4\end{bmatrix}$ and $x = (2,1,1).$

```python
A = np.array([[1, 2, 0],[3, -1, 4]])
x = np.array([2,1,1])
final = 0
for j in range(0,3):
    final += A[1,j]*x[j]
print(final)
```
